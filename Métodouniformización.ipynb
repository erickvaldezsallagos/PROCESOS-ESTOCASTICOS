**EJERCICIO 3:**

**Calcular $P(0.5)$, $P(1)$ y $P(5)$ para la matriz $R$ del ejercicio $1$.**

Importamos las librerias:

```python
import sympy as sp
from math import exp, sqrt, factorial
```

Matriz de tasas R

```python
R = sp.Matrix([
    [0, 2, 3, 0],
    [4, 0, 2, 0],
    [0, 2, 0, 2],
    [1, 0, 3, 0]
])
```

Ahora calculamos $r$

```python
ri = [sum(R.row(i)) for i in range(R.rows)]
r = max(ri)

print("r =", r)
```

Matriz uniforme

```python

n = R.rows
P_hat = sp.zeros(n)

for i in range(n):
    for j in range(n):
        if i == j:
            P_hat[i, j] = 1 - ri[i]/r
        else:
            P_hat[i, j] = R[i, j]/r

print("\nP =")
sp.pprint(P_hat)
```

Definimos la función

```python
def uniformizacion(P_hat, r, t):

    M = int(max(r*t + 5*sqrt(r*t), 20))

    P_t = sp.zeros(P_hat.rows)

    for k in range(M + 1):
        coef = exp(-r*t) * (r*t)**k / factorial(k)
        P_t += coef * (P_hat**k)

    return sp.N(P_t, 8), M
```

De lo anterior, calculamos $P(0.5)$, $P(1)$ y $P(5)$

```python
for t in [0.5, 1, 5]:

    P_t, M = uniformizacion(P_hat, r, t)

    print(f"\n{'='*50}")
    print(f"P({t})")
    print("M =", M)
    sp.pprint(P_t)
```

**3.2 Verificación Chapman-Kolmogorov**

```python
P05, _ = uniformizacion(P_hat, r, 0.5)
P1, _ = uniformizacion(P_hat, r, 1)

CK = sp.N(P05 * P05, 8)

print("\n" + "="*50)
print("P(1) calculada:")
sp.pprint(P1)

print("\nP(0.5)·P(0.5):")
sp.pprint(CK)

print("\nDiferencia:")
sp.pprint(sp.N(P1 - CK, 8))
```

**Conclusión:**

Mediante el método de uniformización se logró aproximar la matriz de transición $P(t)$ de la cadena de Markov en tiempo continuo a partir de la matriz de tasas R. Los resultados obtenidos para $P(0.5)$, $P(1)$ y $P(5)$ muestran la evolución de las probabilidades de transición conforme aumenta el tiempo. Además, la verificación de la ecuación de Chapman-Kolmogorov confirmó la consistencia de los cálculos, ya que $P(1)$ resultó prácticamente igual a $P(0.5)P(0.5)$.

Este teorema se puede usar así: Suponga que se desea calcular P(t)
con una tolerancia ε. Elija M tal que

$\sum_{k=M+1}^{\infty}

\leq \varepsilon.$

SOLUCIÓN POR SIMULACIÓN:

PUNTO 1. Dados $R$, $t$ y $0 < epsilon < 1$

```python
import sympy as sp
```

```python
R = sp.Matrix([
    [0, 2, 3, 0],
    [4, 0, 2, 0],
    [0, 2, 0, 2],
    [1, 0, 3, 0]
])

t = 0.5          # Cambiar por 1 o 5 (DEPENDIENDO)
epsilon = 1e-5
```

PUNTO 2. Calcular r usando la definición

```python
ri = [sum(R.row(i)) for i in range(R.rows)]

r = max(ri)

print("r =", r)
```

PUNTO 3. Calcular P_hat

```python
n = R.rows

P_hat = sp.zeros(n)

for i in range(n):
    for j in range(n):

        if i == j:
            P_hat[i, j] = 1 - ri[i]/r
        else:
            P_hat[i, j] = R[i, j]/r

print("\nP_hat =")
sp.pprint(P_hat)
```

PUNTO 4:

```python
A = P_hat

B = sp.exp(-r*t) * sp.eye(n)

c = sp.exp(-r*t)

suma = c

k = 1
```

PUNTO 5:

```python
while float(suma) < 1 - epsilon:

    c = c*(r*t)/k

    B = B + c*A

    A = A*P_hat

    suma = suma + c

    k = k + 1
```

PUNTO 6:

```python
P_t = B
M = k - 1

print("\nValor de M =", M)

print(f"\nP({t}) =")
sp.pprint(sp.N(P_t, 8))
```

Mediante la implementación del algoritmo de uniformización con una tolerancia de
